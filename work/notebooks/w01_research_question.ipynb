## 1. My lane (or freestyle) and why

**Lane 2: Refresh / Content Opportunity Scoring.**

**Why:** Identifying stale, declining pages provides immediate, measurable business value. Instead of just analyzing data, this lane directly creates a ranked queue that tells a content team exactly where to spend their time to win back lost traffic.

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages should be reviewed first for a content refresh?

**Action:** Route a ranked list of high-priority, declining pages to the writing team to update titles, metadata, or core content.

**Cost of a wrong call:** Wasting writers' time and budget updating pages that either don't need it or have zero search demand, while the pages that actually matter continue to bleed traffic.

## 3. Quick look at the data (2-3 real numbers)



```python
import pandas as pd

# Load the dataset (adjust the path if it gives you an error, usually it's just 'data/raw/...')
df = pd.read_csv("https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
stale_pages = len(df[df['days_since_last_update'] >= 180])
declining_pages = len(df[df['trend_direction'].str.lower() == 'down'])

print(f"1. Total pages in our inventory: {total_pages}")
print(f"2. Pages that are 'stale' (no updates in 6+ months): {stale_pages}")
print(f"3. Pages currently on a downward trend: {declining_pages}")
```
