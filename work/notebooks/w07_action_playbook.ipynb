**Intended Use:** This playbook is a triage tool for SEO and editorial teams to prioritize limited writing resources. It tells them exactly which pages will yield the highest ROI for content updates.

## 3. Human review + the no-go list

**The No-Go List:** Never touch legal, compliance, or privacy policy pages, even if their CTR is terrible. Do not rewrite titles for core branded search terms.
