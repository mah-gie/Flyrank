## 1. My lane as an ML task (type)

This is a Classification task (predicting a binary 1 or 0 for whether a page is declining) that will be used for Ranking (sorting the pages by their probability of decline to create a prioritized queue).

## 2. Target or proxy

The target is whether a page's performance is decaying. Since we can't perfectly measure "decay," our proxy label will be `is_declining_label`, which is a binary 1 or 0 derived from checking if `trend_direction == 'down'`.

## 3. Success metric

The success metric is **Precision@K** (specifically Precision@50 or Precision@20).

**We want to know:** out of the top K pages our model flags for a refresh, what percentage are actually declining?

## 4. The unit of analysis, as a real dataframe



```python
import pandas as pd

# Load up the starter dataset
df = pd.read_csv("https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")

# Create our proxy target column
df['is_declining_label'] = df['trend_direction'].str.lower().eq('down').astype(int)

# Define what one row is, then show a sketch of it
print("Unit of Analysis: 1 row = 1 unique web page.\n")
print("Sketch of the dataframe with our new target column:")

display_columns = ['content_age_days', 'days_since_last_update', 'impressions_90d', 'trend_direction', 'is_declining_label']
print(df[display_columns].head(3))
```

## 5. Why ML beats a fixed rule here

A rigid, hand-written rule (like "age > 180 days AND impressions > 500") is great for finding obvious decay at the very top of the list, but it runs out of signal quickly. ML beats a fixed rule because it can weigh multiple weak signals together (like a slight drop in CTR combined with a specific average position) to find hidden patterns of decay that a simple "yes/no" human rule would completely miss.
